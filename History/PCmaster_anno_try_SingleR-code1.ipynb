```r
seurat_obj = readRDS("Ara_Root_PRJNA640389_GSM4625993_data_R_0_23_5_13.rds")
metadata <- read.csv("ARA_PRJNA640389_celltype.csv")

# 去除Seurat对象中的"cellname"后缀
# rownames(seurat_obj@meta.data) 这里才是细胞名
seurat_obj$cellname <- rownames(seurat_obj@meta.data)

# 去除CSV文件中的"cellname"后缀
metadata$cellname <- sapply(strsplit(metadata$cellname, "-"), function(x) x[[1]])
metadata$cellname <-  paste(metadata$cellname, "-1", sep = "")

# 使用交集操作获取匹配的"cellname"
matching_cellnames <- intersect(seurat_obj$cellname, metadata$cellname)

# 根据匹配的"cellname"筛选metadata和Seurat对象
matching_metadata <- metadata[metadata$cellname %in% matching_cellnames, ]
matching_seurat_obj <- subset(seurat_obj, cells = matching_cellnames)
                            
if( length(rownames(matching_seurat_obj@meta.data)) != length(rownames(matching_metadata)) ){
    print("length not same!")
    # 根据列名找出重复的列值
    duplicated_values <- matching_metadata$cellname[duplicated(matching_metadata$cellname)]
    
    # 根据重复的列值创建逻辑向量，用于筛选要删除的行
    rows_to_remove <- matching_metadata$cellname %in% duplicated_values
    
    # 删除包含重复列值的行
    matching_metadata_no_duplicates <- matching_metadata[!rows_to_remove, ]
    
    # 输出删除重复行后的数据框
    print(matching_metadata_no_duplicates)
    print(length(matching_metadata_no_duplicates$cellname))
    matching_cellnames <- intersect(matching_seurat_obj$cellname, matching_metadata_no_duplicates$cellname)
    
    matching_metadata <- matching_metadata_no_duplicates
    matching_seurat_obj <- subset(matching_seurat_obj, cells = matching_cellnames)
}
                            
# 将"celltype"列中的值添加到Seurat对象的元数据中
tmp = list()
for(i in rownames(matching_seurat_obj@meta.data)){
    tmp = c(tmp,matching_metadata$celltype[matching_metadata$cellname==i])
}
matching_seurat_obj <- AddMetaData(object = matching_seurat_obj, metadata = as.character(tmp), col.name = "celltype")
print(length(matching_seurat_obj$celltype))
```

```r
for(i in 1:10){
    cell_names <- rownames(matching_seurat_obj@meta.data)
    set.seed(i*100)  # 设置随机种子以确保结果可重复
    shuffled_names <- sample(cell_names)
    total_cells <- length(cell_names)
    part1_cells <- round(0.8 * total_cells)
    part2_cells <- round(0.1 * total_cells)
    part3_cells <- total_cells - part1_cells - part2_cells
    part1 <- shuffled_names[1:part1_cells]
    part2 <- shuffled_names[(part1_cells + 1):(part1_cells + part2_cells)]
    part3 <- shuffled_names[(part1_cells + part2_cells + 1):total_cells]
    # print('part1 = ')
    # part1
    # print('part2 = ')
    # part2
    # print('part3 = ')
    # part3
    # print(length(part1))
    # print(length(part2))
    # print(length(part3))
    train_obj <- subset(matching_seurat_obj, cells = part1)
    val_obj <- subset(matching_seurat_obj, cells = part2)
    test_obj <- subset(matching_seurat_obj, cells = part3)
    
    train_obj_count=train_obj[["RNA"]]@counts
    
    pdata=train_obj@meta.data[,c("cellname","celltype")]
    rownames(pdata)=pdata$cellname
    pdata$cellname=NULL
    colnames(pdata)="ref_label"
    
    train_obj_SE <- SummarizedExperiment(assays=list(counts=train_obj_count),colData = pdata)
    train_obj_SE <- logNormCounts(train_obj_SE) #Compute log-transformed normalized expression values
    
    test_obj_count=test_obj[["RNA"]]@counts
    common_gene <- intersect(rownames(test_obj_count), rownames(train_obj_SE))
    train_obj_SE <- train_obj_SE[common_gene,]
    test_obj_count <- test_obj_count[common_gene,]
    
    test_obj_SE <- SummarizedExperiment(assays=list(counts=test_obj_count))
    test_obj_SE <- logNormCounts(test_obj_SE)
    
    singleR_res <- SingleR(test = test_obj_SE, ref = train_obj_SE, labels = train_obj_SE$ref_label)
    anno_df <- as.data.frame(singleR_res$labels)
    anno_df$cellname <- rownames(singleR_res)
    colnames(anno_df)[1] <- 'ref_label_from_train_obj'
    
    test_obj@meta.data=test_obj@meta.data%>%inner_join(anno_df,by="cellname")
    rownames(test_obj@meta.data)=test_obj@meta.data$cellname
    
    count = 0
    for(i in 1:length(test_obj@meta.data$cellname)){
        if(test_obj@meta.data$celltype[i] == test_obj@meta.data$ref_label_from_train_obj[i]){
            count = count + 1
        }
    }
    print(count)
}
```
